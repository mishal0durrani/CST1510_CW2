```python
import hashlib
#input str
plaintext= "DubadduWariWari"
print(type(plaintext))
#convert to binary
binarytext= b"DubadduWariWari"
print(type(binarytext))
#apply hashlib to get hexadecimal hash value
hashtext = hashlib.sha256(binarytext).hexdigest()
print(hashtext)
```

```python
import bcrypt
password = b"password123"
#generate random value
salt = bcrypt.gensalt()
#create hash value
hsh= bcrypt.hashpw(password,salt)
print(hsh)
```

```python
import bcrypt
password = b"password123"
#generate random value
salt1 = bcrypt.gensalt()
salt2 = bcrypt.gensalt()
#create hash value
hash1= bcrypt.hashpw(password,salt1)
hash2= bcrypt.hashpw(password,salt2)
print("Alice:",hash1)
print("Bob:",hash2)
```
