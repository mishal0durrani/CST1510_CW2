```python
import json

text = '{"name": "Rex", "age": 4, "vaccinated": true, "owner": null, "tags": ["dog", "friendly"]}'
print(type(text))                    # just text

pet = json.loads(text)               # this is what response.json() does for you
print(type(pet))                     # a dictionary
print(pet["name"], pet["vaccinated"], pet["owner"], pet["tags"][0])
```

```python
response = requests.get(
    "https://api.github.com/orgs/python/repos",
    params={"per_page": 5, "sort": "updated"},
    timeout=10,
)
for repo in response.json():
    print(repo["name"], repo["open_issues_count"])
    print(repo)
```

```python
data = requests.get("https://api.github.com/orgs/python").json()

# data["repo_count"] would raise KeyError - that field does not exist.
# .get() returns a fallback instead of crashing:
repo_count = data.get("repo_count", "lala")
print(repo_count)

# safe to use the same way even on a field that does exist:
print(data.get("public_repos", "unknown"))
```

```python
import pandas as pd

records = [
    {"id": "a1", "properties": {"mag": 4.5, "place": "Chile", "time": 1700000000000}},
    {"id": "a2", "properties": {"mag": None, "place": "Japan", "time": 1700003600000}},
    {"id": "a3", "properties": {"mag": 5.1, "place": "Peru", "time": 1700007200000}},
]

df = pd.json_normalize(records)
print(df.columns.tolist())
df = pd.json_normalize(records, sep="_")
print(df.columns.tolist())        
```

```python
import re

line = "2026-09-08 connection from 10 0 0 5 refused 10.0.0.6"

match = re.search(r"\d+\.\d+\.\d+\.\d+", line)
print(match.group())
```

```python
no_ip = "no address in this line"
match = re.search(r"\d+\.\d+\.\d+\.\d+", no_ip)
print(match)

if match:
    print(match.group())
else:
    print("no IP found")
```

```python
log = "attempts from 10.0.0.5 and 192.168.1.1 and 10.0.0.9"
all_ips = re.findall(r"\d+\.\d+\.\d+\.\d+", log)
print(all_ips)
```

```python
line = "2026-09-08 connection from 10 0 0 5 refused 10.0.0.6"
date_match = re.search(r"\d{4}-\d{2}-\d{2}", line)
print(date_match.group())

sample = "Supports Python 3.12 and above, tested on 3.10 too"
print(re.findall(r"\d+\.\d+", sample))
```

```python
text = "Server IP: 192.168.0.10, DNS: 8.8.8.8, Gateway: 10.0.0.1"

redacted = re.sub(r"\d+\.\d+\.\d+\.\d+", "[REDACTED]", text)
print(redacted)
```
